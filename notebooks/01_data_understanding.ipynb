## 4. Data Dictionary

A data dictionary provides descriptions of all
variables in the dataset, including their meanings
and data types.

It helps us understand the structure of the data
and choose appropriate analytical techniques.

```python
data_dictionary = pd.DataFrame(
    {
        "Column": df.columns,
        "Data_Type": df.dtypes.values,
        "Unique_Values": [df[column].nunique() for column in df.columns],
        "Missing_Values": [df[column].isnull().sum() for column in df.columns],
    }
)

display(data_dictionary)
```

## 5. Data Validation

In this section, we check whether the dataset
contains logically inconsistent or invalid values.

The objective is to identify data quality issues
before performing cleaning and analysis.

```python
columns_to_check = [
    "Age",
    "MonthlyIncome",
    "DistanceFromHome",
    "TotalWorkingYears",
    "YearsAtCompany",
    "YearsInCurrentRole",
    "YearsSinceLastPromotion",
]

df[columns_to_check].agg(["min", "max"])
```

```python
checks = {
    "YearsAtCompany > TotalWorkingYears": (
        df["YearsAtCompany"] > df["TotalWorkingYears"]
    ).sum(),
    "YearsInCurrentRole > YearsAtCompany": (
        df["YearsInCurrentRole"] > df["YearsAtCompany"]
    ).sum(),
    "YearsSinceLastPromotion > YearsAtCompany": (
        df["YearsSinceLastPromotion"] > df["YearsAtCompany"]
    ).sum(),
    "Negative MonthlyIncome": (df["MonthlyIncome"] < 0).sum(),
    "Negative Age": (df["Age"] < 0).sum(),
}

pd.Series(checks, name="Invalid_Records")
```

```python
categorical_columns = [
    "Attrition",
    "BusinessTravel",
    "Department",
    "EducationField",
    "Gender",
    "JobRole",
    "MaritalStatus",
    "OverTime",
]

for column in categorical_columns:
    print(f"\n{column}")
    print(df[column].value_counts())
```
